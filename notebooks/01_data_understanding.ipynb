```python
import pandas as pd
import numpy as np

# 1. Load raw dataset
file_path = '../data/raw/uber.xlsx'
df = pd.read_excel(file_path, sheet_name='Sheet1')

# 2. Basic dimensions & structure
print("================ DATASET OVERVIEW ================")
print(f"Total Records (Rows): {df.shape[0]:,}")
print(f"Total Features (Cols): {df.shape[1]}")
print("\n--- Column Data Types ---")
print(df.dtypes)

# 3. First 3 records preview
print("\n--- Head Preview ---")
display(df.head(3))
```

```python
# 1. Summary DataFrame (Data Types, Missing Values, Unique Counts)
summary_df = pd.DataFrame({
    'Data Type': df.dtypes,
    'Null Count': df.isnull().sum(),
    'Null %': (df.isnull().sum() / len(df) * 100).round(2),
    'Unique Values': df.nunique()
})

print("================ COLUMN PROFILING ================")
display(summary_df)

# 2. Target Variable Analysis
print("\n================ TARGET VARIABLE (Booking Status) ================")
target_counts = df['Booking Status'].value_counts(dropna=False)
target_pct = (df['Booking Status'].value_counts(normalize=True, dropna=False) * 100).round(2)

target_summary = pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_pct})
display(target_summary)
```

```python
# 1. Check Exact Full-Row Duplicates
full_dups = df.duplicated().sum()

# 2. Check Duplicate Booking IDs
booking_id_dups = df.duplicated(subset=['Booking ID']).sum()

print("================ DUPLICATE VALIDATION ================")
print(f"Full Row Duplicates: {full_dups}")
print(f"Duplicate Booking IDs: {booking_id_dups}")

# 3. If duplicate Booking IDs exist, let's inspect a sample pair
if booking_id_dups > 0:
    print("\n--- Sample Duplicate Booking ID Records ---")
    dup_ids = df[df.duplicated(subset=['Booking ID'], keep=False)]['Booking ID'].head(4).tolist()
    display(df[df['Booking ID'].isin(dup_ids)].sort_values(by='Booking ID'))
```

```python
# Numerical columns range & distribution check
num_cols = ['Booking Value', 'Ride Distance', 'Driver Ratings', 'Customer Rating']

print("================ NUMERICAL RANGE & SANITY CHECK ================")
display(df[num_cols].describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']])

# Check for negative or zero values where not expected
print("\n--- Value Anomaly Checks ---")
print("Negative Fares:", (df['Booking Value'] < 0).sum())
print("Zero Fares:", (df['Booking Value'] == 0).sum())
print("Negative Distances:", (df['Ride Distance'] < 0).sum())
print("Zero Distances:", (df['Ride Distance'] == 0).sum())
print("Ratings out of 1-5 range (Driver):", (~df['Driver Ratings'].dropna().between(1, 5)).sum())
print("Ratings out of 1-5 range (Customer):", (~df['Customer Rating'].dropna().between(1, 5)).sum())
```

```python
print("================ CATEGORICAL INTEGRITY CHECKS ================")

# 1. Unique categories check
print("1. Vehicle Types:", df['Vehicle Type'].unique().tolist())
print("2. Payment Methods:", df['Payment Method'].dropna().unique().tolist())

# 2. Cancellation / Failure Reasons Alignment Check
print("\n--- Failure Reasons Breakdown ---")
print("\n[Rider Cancellation Reasons]:")
print(df['Reason for cancelling by Customer'].value_counts(dropna=False))

print("\n[Driver Cancellation Reasons]:")
print(df['Driver Cancellation Reason'].value_counts(dropna=False))

print("\n[Incomplete Ride Reasons]:")
print(df['Incomplete Rides Reason'].value_counts(dropna=False))
```

```python
# 1. Clean column names to snake_case
df_clean = df.copy()
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('/', '_')
)

# 2. Convert Date to proper string / standard date format
df_clean['date'] = pd.to_datetime(df_clean['date']).dt.date

# 3. Export to processed folder
output_path = '../data/processed/uber_cleaned.csv'
df_clean.to_csv(output_path, index=False)

print("================ DATA CLEANING & EXPORT COMPLETE ================")
print(f"Cleaned dataset saved successfully to: {output_path}")
print(f"Processed Shape: {df_clean.shape}")
print("\nCleaned Column Names:")
print(df_clean.columns.tolist())
```
